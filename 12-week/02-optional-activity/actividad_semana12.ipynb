# Actividad Semana 12 - Rediseño de un gráfico

El objetivo es identificar problemas en una visualización y rediseñarla usando una escala honesta, un título que comunique la conclusión y color con propósito.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

BASE = Path.cwd()
df = pd.read_csv(BASE / "datos_produccion.csv")
df
```

## Gráfico original

El siguiente gráfico se diseñó intencionalmente con problemas: el eje Y está truncado, el título es poco informativo y se usan varios colores sin una razón clara.


```python
plt.figure(figsize=(9, 5.5))
plt.bar(
    df["mes"],
    df["produccion"],
    color=["red", "orange", "purple", "green", "blue", "pink"]
)
plt.ylim(900, 1060)
plt.title("Producción")
plt.xlabel("Mes")
plt.ylabel("Unidades")
plt.tight_layout()
plt.show()
```

## Diagnóstico

El eje vertical comienza en 900 y no en cero, por lo que pequeñas diferencias entre los meses parecen muy grandes. Los colores no representan ninguna categoría ni indicador y generan ruido visual. El título tampoco ayuda a interpretar el resultado porque no expresa cuál es el hallazgo principal.


## Gráfico rediseñado

En el nuevo gráfico la escala inicia en cero, el título comunica la conclusión y el color se usa solamente para destacar el mes con mayor producción.


```python
colores = ["#B8B8B8"] * len(df)
colores[-1] = "#1F77B4"

plt.figure(figsize=(9, 5.5))
plt.bar(df["mes"], df["produccion"], color=colores)
plt.ylim(0, 1100)
plt.title("La producción aumenta y junio registra el valor más alto")
plt.xlabel("Mes")
plt.ylabel("Producción (unidades)")
plt.grid(axis="y", alpha=0.2)
plt.tight_layout()
plt.show()
```

## Justificación

En el rediseño hice que el eje vertical comenzara en cero para mostrar las diferencias de forma más honesta. Cambié el título por uno que comunica directamente la conclusión principal del gráfico. También eliminé los colores innecesarios y dejé un solo color destacado para señalar el mes con mayor producción. Con estos cambios el gráfico queda más claro, fácil de leer y menos propenso a interpretaciones exageradas.
