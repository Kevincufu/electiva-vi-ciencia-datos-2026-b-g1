# Actividad Semana 13 - Diseño de un dashboard

Proceso elegido: **ventas**.

El objetivo es definir 3 KPIs, proponer los gráficos y filtros del tablero y justificar la jerarquía visual.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

BASE = Path.cwd()
df = pd.read_csv(BASE / "datos_dashboard.csv", parse_dates=["fecha"])
df.head()
```

## KPIs

1. **Ingresos totales:** mide el valor generado por las ventas.
2. **Unidades vendidas:** mide el volumen total vendido.
3. **Cumplimiento de meta:** compara las unidades vendidas frente a la meta.


```python
ingresos_totales = df["ingresos"].sum()
unidades_totales = df["unidades_vendidas"].sum()
cumplimiento_meta = unidades_totales / df["meta_unidades"].sum() * 100

print(f"Ingresos totales: ${ingresos_totales:,.0f}")
print(f"Unidades vendidas: {unidades_totales:,.0f}")
print(f"Cumplimiento de meta: {cumplimiento_meta:.1f}%")
```

## Filtros

El dashboard tendrá filtros por:
- Fecha
- Planta
- Producto


```python
mensual = df.groupby("fecha", as_index=False)["ingresos"].sum()

plt.figure(figsize=(9, 4))
plt.plot(mensual["fecha"], mensual["ingresos"], marker="o")
plt.title("Evolución mensual de ingresos")
plt.xlabel("Mes")
plt.ylabel("Ingresos")
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
producto = df.groupby("producto", as_index=False)["unidades_vendidas"].sum()

plt.figure(figsize=(7, 4))
plt.bar(producto["producto"], producto["unidades_vendidas"])
plt.title("Unidades por producto")
plt.xlabel("Producto")
plt.ylabel("Unidades")
plt.ylim(bottom=0)
plt.tight_layout()
plt.show()
```

```python
planta = df.groupby("planta", as_index=False)["unidades_vendidas"].sum()

plt.figure(figsize=(7, 4))
plt.bar(planta["planta"], planta["unidades_vendidas"])
plt.title("Unidades por planta")
plt.xlabel("Planta")
plt.ylabel("Unidades")
plt.ylim(bottom=0)
plt.tight_layout()
plt.show()
```

## Jerarquía visual

Los KPIs se colocan arriba porque resumen el estado del proceso y deben leerse primero. Debajo se muestra la tendencia mensual para entender cómo cambian los ingresos a lo largo del tiempo. Las comparaciones por producto y planta quedan a la derecha porque sirven para profundizar después de revisar el resultado general. Los filtros se ubican antes de los indicadores para cambiar el contexto de análisis fácilmente.
